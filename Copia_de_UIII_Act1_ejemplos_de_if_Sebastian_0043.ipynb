```python
# 1. Ejemplo de if simple con Operadores Relacionales
# Este tipo de estructura ejecuta un bloque de código únicamente si la condición es verdadera
# Ejemplo: Verificar si un número es positivo
numero = 15

if numero > 0: # Utiliza el operador relacional 'mayor que' (>)
    print("El número es positivo.")
```

```python
# 1. Ejemplo de if simple con Operadores Relacionales
# Este tipo de estructura ejecuta un bloque de código únicamente si la condición es verdadera
# Ejemplo: Verificar si un número es positivo
numero = 0

if numero > 0: # Utiliza el operador relacional 'mayor que' (>)
    print("El número es positivo.")
```

```python
# 2. Ejemplo de if-else con Operadores Aritméticos
# La estructura if-else permite elegir entre dos alternativas.
# Aquí utilizaremos el operador aritmético de módulo (%) para obtener el resto de una división.
# Ejemplo: Determinar si un número entero es par o impar
numero = 8

if numero % 2 == 0: # '%' calcula el resto; '==' compara si es igual a 0
    print("El número es par.")
else:
    print("El número es impar.")
```

```python
# 2. Ejemplo de if-else con Operadores Aritméticos
# La estructura if-else permite elegir entre dos alternativas.
# Aquí utilizaremos el operador aritmético de módulo (%) para obtener el resto de una división.
# Ejemplo: Determinar si un número entero es par o impar
numero = 9

if numero % 2 == 0: # '%' calcula el resto; '==' compara si es igual a 0
    print("El número es par.")
else:
    print("El número es impar.")
```

```python
# 3. Ejemplo de if-elif-else con Operadores Lógicos y Aritméticos
# Esta estructura es ideal para manejar múltiples casos.
# En este ejemplo calculamos un descuento basado en la cantidad comprada.
# Ejemplo: Calcular descuento según la cantidad de artículos
cantidad = 12

if cantidad >= 10: # Operador relacional 'mayor o igual que'
    descuento = 0.2 # 20% de descuento
elif cantidad >= 5: # Se evalúa solo si la primera condición falló
    descuento = 0.1 # 10% de descuento
else:
    descuento = 0 # Sin descuento

print(f"Tu descuento es del {descuento * 100}%.")
```

```python
# 4. Ejemplo Complejo: Lógicos, Relacionales y Aritméticos
# Para decisiones más sofisticadas, Python permite combinar múltiples comparaciones usando operadores lógicos como and (y), or (o) y not (no).
# Ejemplo: Identificar si un año es bisiesto
año = 2024

# Un año es bisiesto si es divisible por 4 Y no por 100, O si es divisible por 400
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print(f"{año} es un año bisiesto.")
else:
    print(f"{año} no es un año bisiesto.")
```
